```python
## In python functions we can return multiple values  AND multiple data type values
def multiply(a,b):
    if a<0:
        print("Error: a is negative") ## printinig a string
    else:
        return a*b,a ## returning an integer AND returning two values at a time
```

```python
multiply(-1,3)
print(multiply(2,3))
```

```python
##*args helps in providing multiple values to a function and **kwargs helps in providing multiple key value pairs to a function
def print_de(*args,**kwargs):
    for i in args:
        print(i)
    for key, value in kwargs.items():
        print(f"{key}: {value}")
```

```python
print_de(1,2,3,4,5, name="Akash", age=20)
```
