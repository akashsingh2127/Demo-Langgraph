###Lambda and Map functions


```python
addition=lambda a,b: a+b
print(addition(2,3))
```

```python
numbers=[1,2,3,4,5]
x=list((map(lambda x:x**2,numbers)))## syntax for map is map(function, iterable)
print(x)
```

```python
##multiple iterables for map function
numbers1=[1,2,3,4,5]
numbers2=[6,7,8,9,10]
x=list(map(lambda x, y: x + y, numbers1, numbers2))
print(x)
```

```python
str1=['1','2','3','4','5']
x=list(map(int,str1))## converting string to integer using map function
print(x)
```

###Filter function

```python
def even(num):
    if num%2==0:
        return True
    else:
        return False
lst=[1,2,3,4,5,6,7,8,9]
x=list(filter(even,lst))## filter function is used to filter the data based on some condition
print(x)
```

```python
y=list(filter(lambda x:x>5 and x%2==0,lst)) ## using multiple conditions in lambda function
print(y)
```
