```python
try:
    a=b
except NameError as ex:
    print(ex)    ##it will display the NameError message....... we could have also printed our own msg if wanted
```

```python
## this is dangerous because we have two exceptions in it ZeroDivisionError and NameError but explicitly we handled only ZeroDivision so it only  prints that error and leaves the NameError thing........ so unHandled exceptions can b dangerous 
try:
    r=1/0
    a=b
except ZeroDivisionError as ze:
    print(ze)
```

```python
##the moment one try finds one exception python stops executing try block..  so, in below code u will get zerodivisionerror instead of nameerror
try:  
    r=1/0
    a=b    
except NameError as nm:
    print(nm)
except Exception as ze:
    print(ze)
```

```python
'''else = when there is no exception do it
finally= even exception there or not do it
'''
try: 
    num=int(input("enter a number")) 
    r=1/num
except ValueError as ve:
    print(ve)
except ZeroDivisionError as nm:
    print(nm)
except Exception as ze:
    print(ze)
else:
    print(r)
finally:
    print("it will get printed anyhow")
```
