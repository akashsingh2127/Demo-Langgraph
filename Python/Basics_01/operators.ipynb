```python
## arithmetic operators-[+,-,*,/,//(floor division),%,**(exponential)]
x=15
y = 4

# 1. Addition (+)
print("Addition (x + y):", x + y)

# 2. Subtraction (-)
print("Subtraction (x - y):", x - y)

# 3. Multiplication (*)
print("Multiplication (x * y):", x * y)

# 4. Division (/) - Always returns a float (decimal number)
print("Division (x / y):", x / y)

# 5. Floor Division (//) - Divides and rounds down to the nearest whole number
print("Floor Division (x // y):", x // y)

# 6. Modulus (%) - Returns the remainder of the division
print("Modulus/Remainder (x % y):", x % y)

# 7. Exponentiation (**) - x raised to the power of y (15^4)
print("Exponentiation (x ** y):", x ** y)
```

```python
##Comparison Operators- [==(equal to),!=(not equal to),>,>=(greater than equal to),<(less than),<=(less than equal to)]
# 1. Equal To (==)
x = 5
y = 10
print("Equal to (x == 5):", x == 5)
print("Equal to (x == y):", x == y)

# 2. Not Equal To (!=)
a = "apple"
b = "banana"
print("Not equal to (a != b):", a != b)
print("Not equal to (a != 'apple'):", a != "apple")

# 3. Greater Than (>) and Less Than (<)
score = 85
print("Greater than (score > 50):", score > 50)
print("Less than (score < 50):", score < 50)

# 4. Greater Than or Equal To (>=) and Less Than or Equal To (<=)
age = 18
print("Greater than or equal to (age >= 18):", age >= 18)
print("Less than or equal to (age <= 20):", age <= 20)

# 5. Chaining Comparison Operators
current_age = 25
print("Chained comparison (18 <= current_age < 30):", 18 <= current_age < 30)
```

```python
##Identity Operators
x="akash"
y="akash"
z="kapil"
print(x is y)
print(x is z)
```

```python
## Bitwise Operator-[& (AND), | (OR), ^ (XOR), ~ (NOT), << (Zero-fill left shift), >> (Signed right shift)]
a=2
b=3
print(a&b)
print(a|b)
print(a^b)
print(~a)
print(1<<a)##left shift by (b) position
print(1>>a)##right shift by (a) position

```

```python
##Logical Operator-[and, or, not]
x=True
y=False
print(x and y)
print(x or y)
print(not x)
```
