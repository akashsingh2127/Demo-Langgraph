### LIST


```python
## Lists can store values of any data types , unlike cpp or java\
## traversal in it is same as it was in cpp
my_list = [42, "Hello", 3.14, True, [1, 2, 3]]
for i in my_list:
    print(i)
```

```python
## some of the list methods in list
lst=[1,2,3,4,5]
lst2=[8,9]
print(lst+lst2)
print(lst)
lst.append(6)
print(lst)
lst.insert(2,7)
print(lst)
lst.remove(6)
print(lst)
poped=lst.pop()
print(f"popped element is {poped} and the new list is {lst}")
print("use of ENUMERATE FUNCTION to get index along with the values")
for index,value in enumerate(lst):
    print(index,value)
```

```python
## list comprehension
words=["akash","kapil","vineet","sushmita"]
lst=[len(word) for word in words] 
print(lst)
```

```python
## List comprehension with for (and similarly it happens for the conddtional too)
lst=[1,2,3]
lst2=['a','b']
lst3=[[i,j] for i in lst for j in lst2]
print(lst3)
```
