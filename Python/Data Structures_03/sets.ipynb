```python
st1={1,2,3,4}
st2={3,4,5,6}
print(st1.union(st2))## gives the union but doesn't update the st1
print(st1.intersection(st2))## gives the intersection but doesn't update the st1
print(st1)
st1.intersection_update(st2)##it updates the st1 elements
print(st1)
```
