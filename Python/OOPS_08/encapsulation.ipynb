### ENCAPSULATION :- restricting access of certain variables or functions from derived classes or some object components.

(Note):- 
deafualt way:-public
self.__name=name :-private
self._name=name :-protected



Access Modifier	Accessible inside the Same Class?	Accessible inside Derived (Child) Classes?	Accessible Outside the Class (via Objects)?
public	        Yes	                                Yes                                     	Yes 
protected	    Yes	                                Yes                                     	No 
private     	Yes	                                No	                                        No

```python
## public, private and protected in the same code
class Animal:
    def __init__(self, name):
        self.name = name  # public attribute
        self._species = "Unknown"  # protected attribute
        self.__age = 0  # private attribute

#setters and getters for private attribute
    def set_age(self, age):
        if age >= 0:
            self.__age = age
        else:
            print("Age cannot be negative.")

    def get_age(self):
        return self.__age

    def display_info(self):
        print(f"Name: {self.name}")
        print(f"Species: {self._species}")
        print(f"Age: {self.__age}")

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)
        self._species = "Dog"  # overriding protected attribute
        self.breed = breed  # public attribute

    def display_info(self):
        super().display_info()
        print(f"Breed: {self.breed}")
dog = Dog("Buddy", "Golden Retriever")
dog.set_age(5)
dog.display_info()
```
