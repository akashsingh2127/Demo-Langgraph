### INHERITANCE:
 ### A concept whihc helps a class use another class attributes or methods.... without re defining those again

### Single inheritance

```python
class Car:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    def drive(self,model):
        print(f"U can drive this {self.model}")

    def display_info(self):
        return f"{self.year} {self.make} {self.model}"
```

```python
class Tesla(Car):
    def __init__(self, make, model, year, battery_size):
        super().__init__(make, model, year)## ia m asking the Car class to initialize the make, model, and year attributes of the Tesla object. This is done using the super() function, which allows us to call methods from the parent class (Car) in the child class (Tesla). 
        self.battery_size = battery_size # thisis teh extra attribute in Tesla so i have to initailize it seperately

    def display_info(self):
        return f"{super().display_info()} with a {self.battery_size}-kWh battery"
    def isSelfDriving(self,model):               
        #now the initalized varibale r owned by the child class so u need to access using (self.)
        print(f"the model {self.model} si self driving")
```

```python
car1=Tesla("Tesla", "Model S", 2020, 100)
print(car1.display_info())
print(car1.isSelfDriving("Model S"))  
```

```python
# Child class can inherit the methods of parent class without even having them 
car1.drive("Model S")
```

```python
#The objecvt is created for the Parent class
car2=Car("mahindra","Model A",2026)
print(car2.display_info())
print(car2.model)
```

```python
car2.isSelfDriving("Model S")
# it's a parent class object so it won't be able to access methods of child class hence will throw an error
```

### Multiple Inheritance: one child inheriting from multiple parents

```python
class Animal:
    def __init__(self,name):
        self.name=name
class Pet:
    def __init__(self,owner):
        self.owner=owner
## The child class inheriting from Animal And Pet Class
class Dog(Animal,Pet):
    def __init__(self,name,owner,breed):
        self.breed=breed
        ## it is must to put (self) inside the __init__ but while using super() it is okay
        Animal.__init__(self,name)
        Pet.__init__(self,owner)
    def print_details(self):
        print(f"Dog's name is {self.name} whose owner is {self.owner} and the breed is {self.breed}")


dog1=Dog("Eocky","Akash","German Sheferd")
dog1.print_details()
```

### Multi-Level Inheritance: in sequence order for B- A parent and B child
                                               for C-B parent and C child

                                                         A
                                                         ^
                                                         |
                                                         B
                                                         ^
                                                         |
                                                         C


```python
class GrandParent:
    def __init__(self,grandparent_name):
        self.grandparent_name=grandparent_name
class Parent(GrandParent):
    def __init__(self,parent_name,grandparent_name):
        self.parent_name=parent_name
        GrandParent.__init__(self,grandparent_name)
class Child(Parent):
    def __init__(self,child_name,parent_name,grandparent_name):
        self.child_name=child_name
        Parent.__init__(self,parent_name,grandparent_name)
    def print_details(self):
        print(f"Child's name is {self.child_name} whose parent is {self.parent_name} and grandparent is {self.grandparent_name}")
child1=Child("Arav","Ram","Suresh")
child1.print_details()
```

### Hierarchial Inheritance: one parent multiple child classes

(note):  The child classes can't access eachothers methods directly

```python
class Parent1:
    def __init__(self,parent_name):
        self.parent_name=parent_name
class Child1(Parent1):
    def __init__(self,child_name,parent_name):
        self.child_name=child_name
        Parent1.__init__(self,parent_name)
    def print_details(self):
        print(f"Child's name is {self.child_name} whose parent is {self.parent_name}")
class Child2(Parent1):
    def __init__(self,child_name,parent_name):
        self.child_name=child_name
        Parent1.__init__(self,parent_name)
    def print_details(self):
        print(f"Child's name is {self.child_name} whose parent is {self.parent_name}")

child1=Child1("Arav","Ram")
child2=Child2("Rohan","Ram")
child1.print_details()
child2.print_details()
```
