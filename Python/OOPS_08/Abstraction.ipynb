### ABSTRACTION:
hiding impplementation details from the users of the class...... 
It reduces the complexity a lot. user ko bas pata hoga 0object kya krega ,,, kaise krega it is none of his business...
it provides security and better user experinece

```python
from abc import ABC,abstractmethod
class Vehicle(ABC):
    def concrete_method(self):## this can be inherited by the child class and can be used directly
        print("This is a concrete method in the abstract class.")

    ## u don't pass anything inside the abstract method, it is just a placeholder for the child class to implement it
    @abstractmethod   ## this si the decorator to wrapup the abstract method
    def start_engine(self): 
        pass

    @abstractmethod
    def stop_engine(self):
        pass
class Car(Vehicle):
    ## Compulsory to implement the abstract methods in the child class otherwise it will throw an error
    def start_engine(self):
        print("Car engine started.")

    def stop_engine(self):
        print("Car engine stopped.")

car=Car()
car.concrete_method()  ## calling the concrete method from the abstract class
car.start_engine()  ## calling the implemented abstract method from the child class
car.stop_engine()  ## calling the implemented abstract method from the child class
```
