Class:- It's a blueprint for creating objects...
Instance variiables:- the variables or attributes which r cretaed inside a Constructor which can be accessed by all the objects..
Instance Methods:- these r the functions inside the classes which can be accessed by other objects as well and these methods does some operation for that specific object.



```python
class BankAccount:
    # __init__(...) is used to create a cionstructor which helps in intializing an objects with proper arguments.
    def __init__(self,name,amount):
        self.name=name
        self.amount=amount
    def deposit(self,amount):
        self.amount+=amount
        print(f"ur current amoutn is {self.amount}")
    def withdraw(self,amount):
        if(self.amount<amount):
            print(f"not sufficient amount")
        else:
            self.amount-=amount
            print(f'ur current amount is {self.amount}')
    def printAmount(self):
        print(f"ur current amount is {self.amount}")


acc1=BankAccount("Akash",5000)
acc2=BankAccount("Abhay",10000)
acc1.printAmount()
acc1.withdraw(20000)
acc1.deposit(3000)
acc2.printAmount()



```
