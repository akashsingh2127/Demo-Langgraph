```python
## though i have not created any such file like example.txt but still it will create a new file and write the data in it. because we have used the write mode "w" in the open() function.
with open(("example.txt"),"w") as file:
    file.write("Hello Iam akash")
    file.write("\nI am here to learn about file handling")
```

```python
## use of "r" read function in our file
with open(("example.txt"),"r") as file:
    print(file.readlines())  #.readlines() is used to read the entire file and return a list of lines in the file.

    file.seek(0)# moves cursor back to first line....... otherwise theupcoming (for) wonn't print anything

    for lines in file:
        print(lines.strip())  #.strip() is used to remove the extra spaces and new line character from the output.
```

```python
## use of "a" append function in our file. using append mode "a" we can add new data in our existing file without overwriting the existing data.    because if we use "w" now then try to write something then the entire file will be overwritten by the new text
with open(("example.txt"),"a") as file:
    file.write("\nI am here to learn about file handling")
    file.write("\nI am here to learn about file handling")
```
