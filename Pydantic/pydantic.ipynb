### Pydant :- Data Validation Library

why use Pydantic:-
1. supports dataclass, TypedDict(used in langgraph for creating states)
2. schema validataion and serialization
3. provides structured output


```python
from pydantic import BaseModel,Field
from typing import Optional,List
## inheriting BaseModel
class User(BaseModel):
    id:str=Field(..., description="The unique identifier for the user")#... means that this field is required 
    name:str=Field(default=None, description="The name of the user")#default value is None
    age:Optional[int]=None#Optional means that this field is not required
    cities:List[str]=Field(default=[], description="The list of cities the user has lived in")#default value is an empty list

user=User(id="123",name="John Doe",age=30,cities=["New York","Los Angeles"])
print(user)
```

```python
user2=User(id="456",name=234) ## will throw validation error because Pydantic suppports validation
print(user2)
```
